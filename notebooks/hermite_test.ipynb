```julia
using Pkg
# Pkg.add("Polynomials")
# Pkg.add("LinearAlgebra")

using Polynomials, LinearAlgebra, SpecialPolynomials,Plots, GaussQuadrature
```

```julia

function hermite_projection(x_data, f_data, N)
    H = Polynomials.basis.(Hermite, 0:N-1)

    weights = @. exp(-x_data^2 / 2)

    B = hcat([p.(x_data) for p in H]...)

    W_sqrt = Diagonal(sqrt.(weights))
    B_w = W_sqrt * B
    f_w = W_sqrt * f_data

    c = B_w \ f_w

    sum(c .* H)
end
```

```julia
f(x) = @. (sin(x) + 0.5 * cos(3 * x)+1)*exp(-x^2)
```

```julia

M = 16
x_data = range(0, 4, length=M)
f_data = f(x_data)

poly = hermite_projection(x_data, f_data, 10)

plot(x_data, f_data, label="Data", lw=2)
plot!(poly, minimum(x_data), maximum(x_data), label="Hermite Projection", lw=2)
```

```julia
using CubicSplines
M = 16
x_data = range(0, 4, length=M)
f_data = f(x_data)

poly = CubicSpline(x_data, f_data)

plot(x_data, f_data, label="Data", lw=2)
plot!(0:0.1:maximum(x_data),poly[0:0.1:maximum(x_data)], label="Hermite Projection", lw=2)
```
