```julia
using Polynomials

# Chebyshev nodes of the first kind on [a, b]
function chebyshev_nodes(N::Int, a::Float64=-1.0, b::Float64=1.0)
    k = 0:N-1
    x = cos.((2 .* k .+ 1) .* pi ./ (2N))   # nodes in [-1, 1]
    return 0.5*(a + b) .+ 0.5*(b - a) .* x  # scale to [a, b]
end

# Chebyshev interpolating polynomial
function chebyshev_interpolant(f_vals::Vector{Float64}, nodes::Vector{Float64}, N::Int)
    # Transform nodes from [a,b] to [-1,1] for standard Chebyshev basis
    a, b = nodes[1], nodes[end]
    x_scaled = 2 .* (nodes .- a) ./ (b - a) .- 1

    # Compute Chebyshev coefficients using discrete orthogonality
    c = zeros(N)
    for j in 1:N
        c[j] = 2/N * sum(f_vals .* cos.((j-1) .* (0:N-1) .* pi ./ N))
    end
    c[1] /= 2  # first coefficient halved

    # Construct Chebyshev polynomial in standard power basis
    T = Polynomials.ChebyshevT(c)
    

    return T
end
```

```julia
N = 20
nodes = chebyshev_nodes(N)
f(x) =  x * exp(-x^2*10)

f_vals= f.(nodes)

p = chebyshev_interpolant(f_vals, nodes, N)

x_fine = LinRange(-1, 1, 100)
y_fine = p.(x_fine)

plot(nodes, f_vals)
plot!(x_fine,y_fine)
```
